```python
from scipy.stats import t,probplot
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 3
reps = 10000
nu = 5
```

```python
draws = t.rvs(nu, size=[reps,n])
```

```python
res = probplot(np.ndarray.flatten(draws), plot=plt)
```

```python
res = probplot(np.mean(draws,1), plot=plt)
```
